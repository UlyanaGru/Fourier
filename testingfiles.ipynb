```python
def getdt(data):
    dt = np.diff(data['Time'])
    fig, (ax1,ax2) = plt.subplots(1,2,figsize=(12,4))
    ax1.plot(data['Time'][1:], dt, 'm.-', markersize=3)
    ax1.set_ylabel('Δt (сек)')
    ax1.set_xlabel('Время (t)')
    ax1.grid(True)
    ax2.plot(data['Time'][1:], dt, 'g.-', markersize=3)
    ax2.set_yscale('log')
    ax2.set_xlabel('Время (t)')
    ax2.set_ylabel('Δt (лог. шкала)')
    ax2.grid(True)
    plt.tight_layout()
    plt.show()
```

### there is linear interpolating

```python
def interpol(data, number):
    tstart = data['Time'].iloc[0]
    tend = data['Time'].iloc[-1]
    dt = 0.001

    regular_times = list(np.arange(tstart, tend + dt, dt))
    regular_data = pd.DataFrame({'Time': regular_times})

    for col in data.columns[1:]:
        init_times = data['Time'].values
        init_heights = data[col].values
        interp = []
        idx = 0
        
        for treg in regular_times:
            while idx < len(init_times) - 1 and init_times[idx + 1] < treg:
                idx += 1
            
            # Интерполяция
            if idx == len(init_times) - 1:
                interp.append(init_heights[idx])
            else:
                t1, t2 = init_times[idx], init_times[idx + 1]
                h1, h2 = init_heights[idx], init_heights[idx + 1]
                # Проверка, что treg в интервале
                if t1 <= treg <= t2:
                    h_interp = h1 + (h2 - h1) * (treg - t1) / (t2 - t1)
                else:
                    # Если из-за погрешности treg чуть больше t2
                    h_interp = init_heights[idx]
                interp.append(h_interp)
        
        regular_data[col] = interp

    regular_data.to_csv(f'regular_data{number}.txt', sep=',', index=False, float_format='%.10E')
    return regular_data
```

```python
data120 = pd.read_csv('datafilm/r12/r12smo0.dat',skiprows=1)
interpol(data120,120)
```
